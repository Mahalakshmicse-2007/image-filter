Step 1:
Import the required libraries: OpenCV, NumPy, and Matplotlib.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

Step 2:
Read the input image (e.g., image.jpg).

```python
img = cv2.imread("img.png")
```

Step 3:
Convert the image from BGR to RGB format for display

```python
image2 = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(image2)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

Step 4:
Apply Averaging Filter using cv2.blur()

```python
kernel = np.ones((11,11),np.float32)/121
average = cv2.filter2D(image2,-1,kernel)

plt.imshow(average)
plt.title("Average Filter")
plt.axis("off")
plt.show()
```

Step 5:
Apply Weighted Averaging Filter using a custom kernel with cv2.filter2D().

```python
kernel1 = np.array([[1,2,1],[2,4,2],[1,2,1]])/16
weighted = cv2.filter2D(image2,-1,kernel1)

plt.imshow(weighted)
plt.title("Weighted Average Filter")
plt.axis("off")
plt.show()
```

Step 6:
Apply Gaussian Filter using cv2.GaussianBlur()

```python
gaussian = cv2.GaussianBlur(image2,(33,33),0)

plt.imshow(gaussian)
plt.title("Gaussian Filter")
plt.axis("off")
plt.show()
```

Step 7:
Apply Median Filter using cv2.medianBlur().

```python
median = cv2.medianBlur(image2,13)

plt.imshow(median)
plt.title("Median Filter")
plt.axis("off")
plt.show()
```

Step 8:
Apply Laplacian Sharpening using Kernel with cv2.filter2D().

```python
kernel2 = np.array([[-1,-1,-1],[2,-2,1],[2,1,-1]])
lap_kernel = cv2.filter2D(image2,-1,kernel2)

plt.imshow(lap_kernel)
plt.title("Laplacian Kernel")
plt.axis("off")
plt.show()
```

Step 9:
Convert image to grayscale and apply Laplacian Operator using cv2.Laplacian()

```python
gray = cv2.cvtColor(image2,cv2.COLOR_RGB2GRAY)
laplacian = cv2.Laplacian(gray,cv2.CV_64F)

plt.imshow(laplacian,cmap='gray')
plt.title("Laplacian Operator")
plt.axis("off")
plt.show()
```

Step 10:
Display all filtered images using a grid layout for comparison

```python
plt.figure(figsize=(12,12))

images = [image2,average,weighted,gaussian,median,lap_kernel,laplacian]
titles = ["Original","Average","Weighted Average","Gaussian","Median","Laplacian Kernel","Laplacian Operator"]

for i in range(7):
    plt.subplot(3,3,i+1)
    plt.imshow(images[i],cmap='gray' if i==6 else None)
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```
